```python
import numpy as np
import matplotlib.pyplot as plt

# ==========================================
# Graph for Problem 3 (Algebra vs Physics)
# ==========================================
alg_X = np.array([75, 80, 93, 65, 87, 71, 98, 68, 84, 77])
phys_Y = np.array([82, 78, 86, 72, 91, 80, 95, 72, 89, 74])

# Regression Line: Y = 0.66X + 28.98
line_X = np.linspace(60, 100, 100)
line_Y = 0.6632 * line_X + 28.977

plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.scatter(alg_X, phys_Y, color='blue', label='Student Grades')
plt.plot(line_X, line_Y, color='red', label='Regression Line (Y on X)')
plt.title('Problem 3: Algebra vs Physics Grades')
plt.xlabel('Algebra Grade (X)')
plt.ylabel('Physics Grade (Y)')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.6)

# ==========================================
# Graph for Problem 4 (Quadratic Curve)
# ==========================================
curve_X = np.array([0, 1, 2, 3, 4, 5, 6])
curve_Y = np.array([2.4, 2.1, 3.2, 5.6, 9.3, 14.6, 21.9])

# Quadratic Model: Y = 2.51 - 1.20X + 0.73X^2
smooth_X = np.linspace(0, 7, 100)
smooth_Y = 2.51 - 1.20 * smooth_X + 0.7333 * (smooth_X**2)

plt.subplot(1, 2, 2)
plt.scatter(curve_X, curve_Y, color='green', label='Data Points')
plt.plot(smooth_X, smooth_Y, color='orange', label='Quadratic Fit')
plt.title('Problem 4: Least Squares Curve')
plt.xlabel('X')
plt.ylabel('Y')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.6)

plt.tight_layout()
plt.show()
```
