```python
import numpy as np
import scipy.stats as stats

# Set alpha level
alpha = 0.05

print("="*50)
print("1. Goodness-of-Fit Test")
print("="*50)

print("--- Single Coin Toss (60 tosses) ---")
obs_1 = np.array([37, 23])
exp_1 = np.array([30, 30])
chi2_1, p_1 = stats.chisquare(f_obs=obs_1, f_exp=exp_1)
print(f"Chi-square statistic: {chi2_1:.2f}")
print(f"p-value: {p_1:.4f}")

print("\n--- Three Coins Tossed (240 times) ---")
# Note: Observed sums to 250, Expected sums to 240.
# Calculating manually to bypass scipy's strict sum check.
obs_3 = np.array([24, 108, 95, 23])
exp_3 = np.array([30, 90, 90, 30])

# a. Hypotheses
print("a. Hypotheses:")
print("   H0: The observed frequencies of heads follow the expected distribution (the coins are fair).")
print("   Ha: The observed frequencies of heads do NOT follow the expected distribution.")

# b. Compute chi-square statistic (Calculated manually)
chi2_3 = np.sum((obs_3 - exp_3)**2 / exp_3)
df_3 = len(obs_3) - 1
p_3 = stats.chi2.sf(chi2_3, df_3) # Calculates the p-value based on the manual chi2 stat
crit_val_3 = stats.chi2.ppf(1 - alpha, df_3)

print(f"\nb. Calculation:")
print(f"   Degrees of Freedom: {df_3}")
print(f"   Critical Value (alpha={alpha}): {crit_val_3:.2f}")
print(f"   Chi-square statistic: {chi2_3:.2f}")
print(f"   p-value: {p_3:.4f}")

# c. Decision and interpretation
print("\nc. Decision and Conclusion:")
if chi2_3 > crit_val_3:
    print("   Decision: Reject H0.")
    print("   Conclusion: There is significant evidence to conclude the coins do not follow the expected distribution.")
else:
    print("   Decision: Fail to reject H0.")
    print("   Conclusion: There is not enough evidence to conclude the coins differ from the expected distribution.")


print("\n\n" + "="*50)
print("2. Test of Independence")
print("="*50)
# Data: [[Got Disease, Did Not Get Disease]]
data_2 = np.array([[9, 42],
                   [17, 28]])

# Perform Chi-square test of independence
chi2_2, p_2, df_2, exp_2 = stats.chi2_contingency(data_2, correction=False)

# a. Expected frequencies
print("a. Expected Frequencies:")
print(f"   Vaccinated & Got Disease: {exp_2[0,0]:.2f}")
print(f"   Vaccinated & No Disease: {exp_2[0,1]:.2f}")
print(f"   Not Vaccinated & Got Disease: {exp_2[1,0]:.2f}")
print(f"   Not Vaccinated & No Disease: {exp_2[1,1]:.2f}")

# b. Perform test
crit_val_2 = stats.chi2.ppf(1 - alpha, df_2)
print(f"\nb. Chi-Square Test:")
print("   H0: Vaccination status and disease occurrence are independent.")
print("   Ha: Vaccination status and disease occurrence are dependent (associated).")
print(f"   Chi-square statistic: {chi2_2:.2f}")
print(f"   Critical Value (alpha={alpha}, df={df_2}): {crit_val_2:.2f}")
print(f"   p-value: {p_2:.4f}")

# c. Interpret
print("\nc. Interpretation:")
if chi2_2 > crit_val_2:
    print("   Decision: Reject H0.")
    print("   Conclusion: Vaccination and disease occurrence are NOT independent. There is a significant relationship between being vaccinated and getting the disease.")
else:
    print("   Decision: Fail to reject H0.")
    print("   Conclusion: There is no significant evidence to suggest an association between vaccination and disease occurrence.")


print("\n\n" + "="*50)
print("3. Chi-Square Test in Real-World Context")
print("="*50)
# Data: [[Slept Well, Did Not Sleep Well]]
data_3 = np.array([[44, 10],
                   [81, 35]])

chi2_3_ind, p_3_ind, df_3_ind, exp_3_ind = stats.chi2_contingency(data_3, correction=False)
crit_val_3_ind = stats.chi2.ppf(1 - alpha, df_3_ind)

# a. Perform test
print("a. Chi-Square Test:")
print("   H0: Sleep quality is independent of the type of pill taken.")
print("   Ha: Sleep quality depends on the type of pill taken.")
print(f"   Expected Frequencies:\n{np.round(exp_3_ind, 2)}")
print(f"   Chi-square statistic: {chi2_3_ind:.2f}")
print(f"   Critical Value (alpha={alpha}, df={df_3_ind}): {crit_val_3_ind:.2f}")
print(f"   p-value: {p_3_ind:.4f}")

# b. Practical interpretation
print("\nb. Practical Interpretation:")
if chi2_3_ind > crit_val_3_ind:
    print("   Decision: Reject H0.")
    print("   Conclusion: There is a statistically significant association between the type of pill taken and how well the patients slept. The sleeping pills had a different effect compared to the sugar pills.")
else:
    print("   Decision: Fail to reject H0.")
    print("   Conclusion: There is no statistically significant difference in sleep quality between those who took sleeping pills and those who took sugar pills. The sleeping pills do not appear to be effective.")


print("\n\n" + "="*50)
print("4. Measure of Association")
print("="*50)
# Data: [[Blonde, Not Blonde]]
data_4 = np.array([[49, 25],
                   [30, 96]])

chi2_4, p_4, df_4, exp_4 = stats.chi2_contingency(data_4, correction=False)
N = np.sum(data_4)

# a. Compute coefficient of contingency
# Formula: C = sqrt(chi2 / (chi2 + N))
C = np.sqrt(chi2_4 / (chi2_4 + N))

print("a. Calculation:")
print(f"   Chi-square statistic: {chi2_4:.2f}")
print(f"   Total sample size (N): {N}")
print(f"   Coefficient of contingency (C): {C:.2f}")

# b. Interpret strength
print("\nb. Interpretation:")
# For a 2x2 table, the maximum value of C is sqrt(1/2) ≈ 0.707
C_max = np.sqrt(1/2)
C_adjusted = C / C_max
print(f"   Maximum possible C for a 2x2 table is approx {C_max:.2f}.")
print(f"   Since {C:.2f} is moderately high (about {(C_adjusted*100):.1f}% of the maximum possible value),")
print("   this indicates a moderate to strong association between eye color (Blue/Not Blue) and hair color (Blonde/Not Blonde) among the students.")
```
