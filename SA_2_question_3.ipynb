1. Multiple Regression Equation

a. Find the least-squares regression equation of $X_3$ on $X_1$ and $X_2$.The multiple regression model is: $\hat{X}_3 = b_0 + b_1X_1 + b_2X_2$Step 1: Compute $b_1$ and $b_2$Formulas:$$b_1 = \frac{(\sum x_2^2)(\sum x_1 x_3) - (\sum x_1 x_2)(\sum x_2 x_3)}{(\sum x_1^2)(\sum x_2^2) - (\sum x_1 x_2)^2}$$$$b_2 = \frac{(\sum x_1^2)(\sum x_2 x_3) - (\sum x_1 x_2)(\sum x_1 x_3)}{(\sum x_1^2)(\sum x_2^2) - (\sum x_1 x_2)^2}$$Substitutions:$$b_1 = \frac{(140)(-582) - (-100)(720)}{(90)(140) - (-100)^2} = \frac{-81480 + 72000}{12600 - 10000} = \frac{-9480}{2600} = -3.646$$$$b_2 = \frac{(90)(720) - (-100)(-582)}{2600} = \frac{64800 - 58200}{2600} = \frac{6600}{2600} = 2.538$$Step 2: Compute $b_0$Formula: $b_0 = \bar{X}_3 - b_1\bar{X}_1 - b_2\bar{X}_2$Substitution: $b_0 = 50 - (-3.646)(8) - (2.538)(7) = 50 + 29.168 - 17.766 = 61.402$Final Equation:$\hat{X}_3 = 61.40 - 3.65X_1 + 2.54X_2$b. Estimate the value of $X_3$ when $X_1 = 10$ and $X_2 = 6$.Substitution: $\hat{X}_3 = 61.402 - 3.646(10) + 2.538(6)$$\hat{X}_3 = 61.402 - 36.46 + 15.228$Estimated $X_3$: 40.17

2. Standard Error of Estimate

a. Compute the standard error of estimate ($s_e$).Formula: $s_e = \sqrt{\frac{SSE}{n - k - 1}}$  where $SSE = \sum x_3^2 - (b_1 \sum x_1 x_3 + b_2 \sum x_2 x_3)$ and $k=2$ independent variables.Step 1: Compute Regression Sum of Squares (SSR) and Error Sum of Squares (SSE)$$SSR = (-3.646)(-582) + (2.538)(720) = 2121.972 + 1827.36 = 3949.332$$(Note: Using exact fractions yields an SSR of exactly 3949.75)$$SSE = SST - SSR = 4008 - 3949.75 = 58.25$$Step 2: Compute $s_e$Substitution:$$s_e = \sqrt{\frac{58.25}{6 - 2 - 1}} = \sqrt{\frac{58.25}{3}} = \sqrt{19.417}$$$s_e$: 4.41
